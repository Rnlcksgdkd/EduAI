```python
# 프롬프트 템플릿
choice_eval_template = PromptTemplate.from_template("""

당신은 객관식 시험 문제의 선택지를 검토하는 전문가입니다.
주어진 문제와 선택지 각각이 정답인지 (논리적으로 맞는 설명인지, 개념적으로 정확한지를) 판단하세요.

각 선택지에 대해 아래 형식으로 정답 가능성을 평가하세요:
[선택지 번호]: [정답 가능성 수치 (0 = 명확한 오답, 1 = 명확한 정답)]

문제: {question}
선택지:
{choice_1}
{choice_2}
{choice_3}
{choice_4}


출력 형식 예시:
1: 0.8
2: 0.2
3: 1
4: 0
""")
```

```python
from langchain.chat_models import ChatOpenAI
from langchain.prompts import PromptTemplate
from langchain.schema.runnable import RunnableLambda, RunnableMap

# LLM
llm = ChatOpenAI(temperature=0)

# 체인
choice_eval_chain = choice_eval_template | llm

# 예시 문제
sample = {
        "question": "합성곱 신경망(CNN)이 특히 강점을 보이는 분야는 무엇인가?",
        "choice_1": "자연어 처리",
        "choice_2": "음성 인식",
        "choice_3": "이미지 분류",
        "choice_4": "시계열 데이터 분석"
}

# 실행
response = choice_eval_chain.invoke(sample)
print(response.content)
```
